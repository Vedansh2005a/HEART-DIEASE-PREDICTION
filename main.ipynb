```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
```

```python
data=pd.read_csv("heart_disease_data.csv")
data.describe()
```

```python
data.info()
```

```python
df=pd.DataFrame(data)
```

1-->DEFECTIVE HEART
0-->GOOD HEART

```python
model=LogisticRegression()

X=df.drop(columns="target",axis=1)
Y=df["target"]
print(X)
print(Y)
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42,stratify=Y)
```

```python
regresd=model.fit(X_train,Y_train)
```

```python
from sklearn.metrics import accuracy_score
X_train_predict=model.predict(X_train)
trained_model_accuracy=accuracy_score(X_train_predict,Y_train)
print(trained_model_accuracy)
```

```python
X_test_predict=model.predict(X_test)
test_model_accuracy=accuracy_score(X_test_predict,Y_test)
print(test_model_accuracy)
```
